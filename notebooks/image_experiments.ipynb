```python
from model_api import ModelAPI
import my_signatures as sig
from textwrap import dedent
import utils
model = ModelAPI()

s = sig.Signature(
    [sig.Field("descriptors", list, ""), sig.Field("image1", utils.imageurl, ""), sig.Field("image2", utils.imageurl, "")],
    [sig.Field("verdict", str, "image1 or image2")],
    dedent("""\
           Compare the two images based on the descriptors provided. 
           Determine which image better represents the descriptors.
           """)
)


model.chain_of_thought(s, descriptors=["current", "formal", "bright"], image1="https://pbs.twimg.com/media/GqPHJG-XgAI5rco?format=jpg&name=small", image2="https://pbs.twimg.com/media/GqPHJHAXkAAkjDT?format=jpg&name=small")
```
