```python
import json
import matplotlib.pyplot as plt

with open('runs/IMAGE/prompts.jsonl') as f:
    prompts = [json.loads(line) for line in f]
with open('runs/IMAGE/task_attempts.json') as f:
    task_attempts = json.load(f)["dev"][0]["attempts"]
```

```python
by_gen = [list(filter(lambda p: p["gen"] == i, prompts)) for i in range(4)]
by_gen_ids = [list(map(lambda p: p["id"], by_gen[i])) for i in range(4)]
```

```python
url_by_gen = [list(filter(lambda x: x["prompt_id"] in gen_ids, task_attempts)) for gen_ids in by_gen_ids]
```

```python
import requests
for i, attempts in enumerate(url_by_gen):
    print(f"Gen {i}")
    for a in attempts:
        url = a['answer']
        response = requests.get(url)
        pid = a['prompt_id'][:5]
        if response.status_code == 200:
            with open(f"runs/IMAGE/images/image-gen{i}-id{pid}.jpg", "wb") as f:
                f.write(response.content)
        else:
            print("Failed to download image:", response.status_code)
```
