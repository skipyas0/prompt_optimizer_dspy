```python
import json
path = 'finished_experiments/LIMERICK/'

with open(path+"prompts.jsonl", 'r') as f:
    prompts = [json.loads(line) for line in f.readlines()] 

with open(path+"task_attempts.json", 'r') as f:
    attempts = json.load(f)
```

```python
gens = [[p for p in prompts if p["gen"] == gen] for gen in range(4)]
last_gen_ids = [p["id"] for p in gens[-1]]
best = [max(g, key=lambda p: p["dev_score"]) for g in gens]
```

```python
for p in best:
    print(f'Best in gen {p["gen"]}: {p["id"]}')
```

```python
q = attempts["dev"][1]
best_ids = [p["id"] for  p in best]
q_attempts = q["attempts"]
best_attempts = list(filter(lambda a: a["prompt_id"] in best_ids, q_attempts))
last_attempts = list(filter(lambda a: a["prompt_id"] in last_gen_ids, q_attempts))
```

```python
for a in best_attempts:
    print(a["answer"]["solution"])
    print("---------")
```
